# Product Master Generation

```python
import pandas as pd
import numpy as np
import random

# Reproducibility
random.seed(42)
np.random.seed(42)

# -----------------------------
# CATEGORY CONFIGURATION
# -----------------------------

category_config = {
    "Biscuits": {
        "count": 40,
        "brands": ["CrispBite", "GoldenCrunch", "DailyTreat"],
        "sizes": ["50g", "100g", "200g"],
        "mrp_range": (5, 50),
        "margin_range": (8, 12),
        "shelf_life": 180
    },
    "Snacks": {
        "count": 45,
        "brands": ["CrunchyKing", "SpiceMunch", "TastyWave"],
        "sizes": ["30g", "60g", "150g"],
        "mrp_range": (10, 60),
        "margin_range": (10, 15),
        "shelf_life": 120
    },
    "Edible Oil": {
        "count": 25,
        "brands": ["PureDrop", "HealthGold", "NaturePress"],
        "sizes": ["500ml", "1L", "5L"],
        "mrp_range": (90, 220),
        "margin_range": (5, 8),
        "shelf_life": 365
    },
    "Personal Care": {
        "count": 35,
        "brands": ["FreshGlow", "CleanAura", "SilkTouch"],
        "sizes": ["75g", "180ml", "250ml"],
        "mrp_range": (20, 180),
        "margin_range": (12, 20),
        "shelf_life": 730
    },
    "Beverages": {
        "count": 25,
        "brands": ["CoolSip", "FreshFizz", "NaturalBurst"],
        "sizes": ["250ml", "500ml", "1L"],
        "mrp_range": (20, 120),
        "margin_range": (8, 15),
        "shelf_life": 150
    }
}

# -----------------------------
# MOVEMENT DISTRIBUTION
# -----------------------------

total_skus = 170
movement_distribution = (
    ["Fast"] * 34 +
    ["Medium"] * 85 +
    ["Slow"] * 51
)

random.shuffle(movement_distribution)

# -----------------------------
# GENERATION LOGIC
# -----------------------------

products = []
product_id = 1
movement_index = 0

for category, config in category_config.items():
    for _ in range(config["count"]):
        brand = random.choice(config["brands"])
        size = random.choice(config["sizes"])

        mrp = round(random.uniform(*config["mrp_range"]), 2)
        margin = round(random.uniform(*config["margin_range"]), 2)

        purchase_price = round(mrp * (1 - margin / 100), 2)

        product_name = f"{brand} {category} {size}"

        products.append([
            product_id,
            product_name,
            category,
            brand,
            size,
            mrp,
            purchase_price,
            margin,
            config["shelf_life"],
            movement_distribution[movement_index]
        ])

        product_id += 1
        movement_index += 1

# -----------------------------
# CREATE DATAFRAME
# -----------------------------

columns = [
    "product_id",
    "product_name",
    "category",
    "brand",
    "unit_size",
    "mrp",
    "purchase_price",
    "margin_percent",
    "shelf_life_days",
    "movement_type"
]

product_master_df = pd.DataFrame(products, columns=columns)

# Save to CSV
product_master_df.to_csv("product_master.csv", index=False)

print("Product master generated successfully.")
print(product_master_df.head())
```

# 2. Supplier Master Generation


```python
import pandas as pd
import random

random.seed(42)
```

```python
brands = [
    "CrispBite", "GoldenCrunch", "DailyTreat",
    "CrunchyKing", "SpiceMunch", "TastyWave",
    "PureDrop", "HealthGold", "NaturePress",
    "FreshGlow", "CleanAura", "SilkTouch",
    "CoolSip", "FreshFizz", "NaturalBurst"
]

cities = ["Delhi", "Gurugram", "Noida", "Ghaziabad", "Faridabad", "Jaipur"]

supplier_data = []
supplier_id = 1

for brand in brands:
    
    # Larger brands get longer credit
    if brand in ["CrispBite", "GoldenCrunch", "PureDrop", "HealthGold", "FreshGlow", "CoolSip"]:
        credit_days = random.randint(21, 30)
    else:
        credit_days = random.randint(14, 21)

    supplier_name = f"{brand} Distributors Pvt Ltd"
    city = random.choice(cities)

    supplier_data.append([
        supplier_id,
        supplier_name,
        brand,
        credit_days,
        city
    ])

    supplier_id += 1
```

```python
supplier_master_df = pd.DataFrame(
    supplier_data,
    columns=[
        "supplier_id",
        "supplier_name",
        "brand",
        "credit_days",
        "city"
    ]
)

supplier_master_df
```

```python
supplier_master_df.to_csv("supplier_master.csv", index=False)
print("Supplier master saved successfully.")
```

#  Retailer Master

```python
import random

random.seed(42)

total_retailers = 220

retailer_distribution = {
    "Kirana": 132,
    "Supermarket": 66,
    "Wholesaler": 22
}

city_distribution = {
    "Delhi": 77,        # 35%
    "Gurugram": 44,     # 20%
    "Noida": 33,        # 15%
    "Ghaziabad": 22,    # 10%
    "Faridabad": 22,    # 10%
    "Jaipur": 22        # 10%
}
```

```python
retailer_data = []
retailer_id = 1

# Create expanded city list based on distribution
city_list = []
for city, count in city_distribution.items():
    city_list.extend([city] * count)

random.shuffle(city_list)

# Create expanded retailer type list
type_list = []
for rtype, count in retailer_distribution.items():
    type_list.extend([rtype] * count)

random.shuffle(type_list)

# Generate retailer records
for i in range(total_retailers):
    
    city = city_list[i]
    retailer_type = type_list[i]
    
    retailer_name = f"{city}_{retailer_type}_{retailer_id}"
    
    retailer_data.append([
        retailer_id,
        retailer_name,
        city,
        retailer_type
    ])
    
    retailer_id += 1
```

```python
import pandas as pd

retailer_master_df = pd.DataFrame(
    retailer_data,
    columns=[
        "retailer_id",
        "retailer_name",
        "city",
        "retailer_type"
    ]
)

retailer_master_df.head()
```

```python
retailer_master_df.to_csv("retailer_master.csv", index=False)
print("Retailer master saved successfully.")
```

#  Purchase Transaction Generation 

```python
import pandas as pd
import numpy as np
import random
from datetime import timedelta

random.seed(42)
np.random.seed(42)

product_master_df = pd.read_csv("product_master.csv")
supplier_master_df = pd.read_csv("supplier_master.csv")

product_master_df.head()
```

```python
brand_supplier_map = supplier_master_df.set_index("brand")["supplier_id"].to_dict()

brand_supplier_map
```

```python
total_skus = len(product_master_df)

# 10% over-purchase SKUs
overpurchase_count = int(total_skus * 0.10)

# 5% stockout SKUs
stockout_count = int(total_skus * 0.05)

all_product_ids = product_master_df["product_id"].tolist()

overpurchase_skus = random.sample(all_product_ids, overpurchase_count)

remaining_skus = list(set(all_product_ids) - set(overpurchase_skus))

stockout_skus = random.sample(remaining_skus, stockout_count)

print("Total SKUs:", total_skus)
print("Overpurchase SKUs:", len(overpurchase_skus))
print("Stockout SKUs:", len(stockout_skus))
```

```python
start_date = pd.to_datetime("2023-04-01")
end_date = pd.to_datetime("2024-03-31")
```

```python
purchase_data = []
purchase_id = 1

for _, row in product_master_df.iterrows():
    
    product_id = row["product_id"]
    brand = row["brand"]
    mrp = row["mrp"]
    movement = row["movement_type"]
    shelf_life = row["shelf_life_days"]
    
    supplier_id = brand_supplier_map[brand]
    
    current_date = start_date
    
    # Determine purchase gap based on movement
    if movement == "Fast":
        min_gap, max_gap = 18, 25
        base_qty = 400
    elif movement == "Medium":
        min_gap, max_gap = 35, 50
        base_qty = 600
    else:  # Slow
        min_gap, max_gap = 65, 95
        base_qty = 1000
    
    while current_date <= end_date:
        
        # Price factor logic
        if mrp <= 30:
            price_factor = 1.3
        elif mrp <= 100:
            price_factor = 1.0
        else:
            price_factor = 0.7
        
        quantity = base_qty * price_factor
        
        # Controlled randomness ±15%
        quantity = quantity * random.uniform(0.85, 1.15)
        
        # Apply imperfection adjustments
        if product_id in overpurchase_skus:
            quantity *= random.uniform(1.25, 1.40)
        
        if product_id in stockout_skus:
            quantity *= random.uniform(0.60, 0.80)
        
        quantity = int(quantity)
        
        # Batch number format: PRODID-YYYYMMDD
        batch_number = f"B{product_id}_{current_date.strftime('%Y%m%d')}"
        
        expiry_date = current_date + timedelta(days=int(shelf_life))
        
        purchase_data.append([
            purchase_id,
            product_id,
            supplier_id,
            current_date,
            batch_number,
            expiry_date,
            quantity,
            row["purchase_price"]
        ])
        
        purchase_id += 1
        
        # Move to next purchase cycle
        gap_days = random.randint(min_gap, max_gap)
        current_date += timedelta(days=gap_days)

purchase_df = pd.DataFrame(
    purchase_data,
    columns=[
        "purchase_id",
        "product_id",
        "supplier_id",
        "purchase_date",
        "batch_number",
        "expiry_date",
        "quantity_purchased",
        "purchase_price"
    ]
)

purchase_df.head()
```

```python
purchase_df.to_csv("purchase_transactions.csv", index=False)
print("Purchase transactions saved successfully.")
```

# Create Inventory Ledger from Purchases

```python
# Sort purchases properly
purchase_df = purchase_df.sort_values(by=["product_id", "purchase_date"])

inventory = {}

for _, row in purchase_df.iterrows():
    
    product_id = row["product_id"]
    
    batch_info = {
        "batch_number": row["batch_number"],
        "remaining_qty": row["quantity_purchased"],
        "expiry_date": pd.to_datetime(row["expiry_date"]),
        "purchase_date": pd.to_datetime(row["purchase_date"])
    }
    
    if product_id not in inventory:
        inventory[product_id] = []
    
    inventory[product_id].append(batch_info)

print("Inventory ledger created.")
```

```python
# Generate all Mondays between start_date and end_date
weekly_dates = pd.date_range(start=start_date, end=end_date, freq="W-MON")

len(weekly_dates), weekly_dates[:3]
```

```python
# Separate retailers by type
kirana_retailers = retailer_master_df[retailer_master_df["retailer_type"] == "Kirana"]["retailer_id"].tolist()
supermarket_retailers = retailer_master_df[retailer_master_df["retailer_type"] == "Supermarket"]["retailer_id"].tolist()
wholesaler_retailers = retailer_master_df[retailer_master_df["retailer_type"] == "Wholesaler"]["retailer_id"].tolist()

print(len(kirana_retailers), len(supermarket_retailers), len(wholesaler_retailers))
```

```python
def choose_retailer():
    r = random.random()
    
    if r <= 0.60:
        return random.choice(kirana_retailers), "Kirana"
    elif r <= 0.90:
        return random.choice(supermarket_retailers), "Supermarket"
    else:
        return random.choice(wholesaler_retailers), "Wholesaler"
```

# Weekly Sales Simulation (FIFO + Expiry + Stock Constraint)

```python
sales_data = []
sales_id = 1

for current_week in weekly_dates:
    
    for _, product in product_master_df.iterrows():
        
        product_id = product["product_id"]
        mrp = product["mrp"]
        movement = product["movement_type"]
        
        # Skip if no inventory exists
        if product_id not in inventory:
            continue
        
        # Remove expired batches first
        inventory[product_id] = [
            batch for batch in inventory[product_id]
            if batch["expiry_date"] >= current_week
        ]
        
        # If no sellable inventory
        if len(inventory[product_id]) == 0:
            continue
        
        # Sale probability based on movement
        if movement == "Fast":
            sale_prob = random.uniform(0.85, 0.95)
            base_weekly_qty = random.randint(80, 150)
        elif movement == "Medium":
            sale_prob = random.uniform(0.55, 0.70)
            base_weekly_qty = random.randint(40, 100)
        else:
            sale_prob = random.uniform(0.25, 0.40)
            base_weekly_qty = random.randint(10, 60)
        
        # Trigger sale?
        if random.random() > sale_prob:
            continue
        
        # Retailer selection
        retailer_id, retailer_type = choose_retailer()
        
        # Retailer multiplier
        if retailer_type == "Kirana":
            multiplier = 1
        elif retailer_type == "Supermarket":
            multiplier = 1.8
        else:
            multiplier = 3
        
        sale_qty = int(base_weekly_qty * multiplier)
        
        # Adjust demand for stockout SKUs
        if product_id in stockout_skus:
            sale_qty = int(sale_qty * random.uniform(1.2, 1.4))
        
        # Total available inventory
        total_available = sum(batch["remaining_qty"] for batch in inventory[product_id])
        
        if total_available == 0:
            continue
        
        # Cap sale quantity to available stock
        sale_qty = min(sale_qty, total_available)
        
        # FIFO deduction
        qty_to_deduct = sale_qty
        
        for batch in inventory[product_id]:
            if qty_to_deduct <= 0:
                break
            
            if batch["remaining_qty"] >= qty_to_deduct:
                batch["remaining_qty"] -= qty_to_deduct
                qty_to_deduct = 0
            else:
                qty_to_deduct -= batch["remaining_qty"]
                batch["remaining_qty"] = 0
        
        # Remove fully depleted batches
        inventory[product_id] = [
            batch for batch in inventory[product_id]
            if batch["remaining_qty"] > 0
        ]
        
        # Selling price logic (Distributor pricing)
        selling_price = round(mrp * random.uniform(0.92, 0.97), 2)
        
        sales_data.append([
            sales_id,
            product_id,
            current_week,
            retailer_id,
            sale_qty,
            selling_price
        ])
        
        sales_id += 1

sales_df = pd.DataFrame(
    sales_data,
    columns=[
        "sales_id",
        "product_id",
        "sale_date",
        "retailer_id",
        "quantity_sold",
        "selling_price"
    ]
)

sales_df.head()
```
